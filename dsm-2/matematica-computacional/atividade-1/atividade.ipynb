```python
import matplotlib.pyplot as plt
numero = input("digite uma sequencia de numeros:")
numeros = [float(x) for x in numero.split()]
print("sua sequencia",numeros)
```

```python
maior = numeros[0]
menor = numeros[0]
soma = 0

for numero in numeros: 
    soma = soma + numero
    
    if numero > maior:
        maior = numero
    
    if numero < menor:
        menor = numero
```

```python
posicaoMaior = []
posicaoMenor = []

for i in range(len(numeros)):
    if numeros[i] == maior:
        posicaoMaior.append(i)
    
    if numeros[i] == menor:
        posicaoMenor.append(i)
```

```python
media = soma / len(numeros)

print("Maior valor:", maior)
print("Menor valor:", menor)
print("Média:", media)
```

```python
plt.plot(numeros)

plt.scatter(
    posicaoMaior,
    [maior] * len(posicaoMaior),
    color="red",
    marker="x",
    s=100
)

plt.scatter(
    posicaoMenor,
    [menor] * len(posicaoMenor),
    color="blue",
    marker="o",
    s=100
)

plt.xlabel("Posição")
plt.ylabel("Valor")
plt.grid(True)

plt.show()
```
